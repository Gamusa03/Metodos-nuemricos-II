# Integración de Romberg

## Definimos la función de la integración 

### Usando $\int_{a}^{b} f(x) \, dx \approx \frac{b - a}{2} \left[ f(a) + f(b) \right]$ para el primer $K\left[k,0\right]$

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def integracion_romberg(f, a, b, n, tabla: bool = True):
    
    if n <= 0:
        raise ValueError("El número debe ser entero positivo.")
        
    # Hacemos la tabla
    R = np.full((n, n), np.nan, dtype=np.float64)
    
    # Regla del trapecio
    h = float(b - a)
    R[0, 0] = 0.5 * h * (f(a) + f(b))
    
    for i in range(1, n):
        
        h /= 2.0
        
        puntos_nuevos = 2**(i - 1)
        puntos = a + (2 * np.arange(1, puntos_nuevos + 1) - 1) * h
        suma_puntos = np.sum(f(puntos))
        
        R[i, 0] = 0.5 * R[i - 1, 0] + h * suma_puntos
        
        for j in range(1, i + 1):
            factor_4 = 4.0**j
            R[i, j] = R[i, j - 1] + (R[i, j - 1] - R[i - 1, j - 1]) / (factor_4 - 1.0)
            
    if tabla:
        columnas = [f"R[k,{j}]" for j in range(n)]
        filas = [f"k={i+1}" for i in range(n)]
        df_tabla = pd.DataFrame(R, index=filas, columns=columnas)
        print(df_tabla.dropna(how='all', axis=1).fillna(''))
        
    return R
```

## Ejemplificando  con $\int_0^{\pi} \sin(x) \ dx$

```python
f = np.sin
a, b = 0.0, np.pi
n = 4

matriz_romberg = integracion_romberg(f, a, b, n)
print(f"\nAproximación (R[{n},{n}]): {matriz_romberg[-1, -1]:.10f}")
```
