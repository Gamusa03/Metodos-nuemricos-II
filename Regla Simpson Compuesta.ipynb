# Regla de Simpson Compuesta

### Vamos a definr la función para $\frac{h}{3}[f(a)+2\sum_{j=1}^{\frac{n}{2}-1}f(x_{2j})+4\sum_{j=1}^{\frac{n}{2}}f(x_{2j-1})+f(b)]$

```python
import math
def simpson_compuesta (n, a, b, f):
    if n<=0 or n% 2!=0:
        raise ValueError("El numero no es par") #Por si ponene un numero que o es par

    #Definimos los valores h y las x´s en cero
    h=(b-a)/n
    x_1=0
    x_2=0

#Hacemos el ciclo para las sumatorias 
    for i in range(1, n):
        x=a+(i*h)
        if i%2==0:
            x_1+=f(x)
        else:
            x_2+=f(x)

    I= (h*(f(a)+2*x_1+4*x_2+f(b)))/3
    return(I)
```

### Definimos $f(x)$ de la integral a resolver

```python
import math
def f(x):
    return math.exp(2*x)* math.sin(3*x)
```

### Llamamos a la función Simpson 

#### Como ejemplo ocuparemos $\int_{0}^{2}e^{2x}\sin{3x}\ dx$

```python
aprox=simpson_compuesta(100,0,2,f)
print(aprox)
```
